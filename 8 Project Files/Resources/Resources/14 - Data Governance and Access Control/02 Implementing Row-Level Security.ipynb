```sql
SELECT
    seller_state,
    COUNT(*) AS seller_count
FROM course_training_catalog.silver_olist.sellers
GROUP BY seller_state
ORDER BY seller_count DESC;
```

```sql
CREATE OR REPLACE FUNCTION
course_training_catalog.silver_olist.filter_seller_state(
    state STRING
)
RETURNS BOOLEAN
RETURN
    CASE 
        WHEN is_account_group_member('southeast_sales_team')
            AND state IN ('SP', 'RJ')
        THEN TRUE 

        WHEN is_account_group_member('south_sales_team')
            AND state IN ('PR', 'SC')
        THEN TRUE 

        ELSE FALSE 

    END;
```

```sql
%sql
SELECT
    course_training_catalog.silver_olist.filter_seller_state('SP') AS sp_access,
    course_training_catalog.silver_olist.filter_seller_state('RJ') AS rj_access,
    course_training_catalog.silver_olist.filter_seller_state('PR') AS pr_access,
    course_training_catalog.silver_olist.filter_seller_state('SC') AS sc_access;
```

```sql
ALTER TABLE course_training_catalog.silver_olist.sellers
SET ROW FILTER course_training_catalog.silver_olist.filter_seller_state
ON (seller_state);
```

```sql
SELECT
    seller_state,
    COUNT(*) AS seller_count
FROM course_training_catalog.silver_olist.sellers
GROUP BY seller_state
ORDER BY seller_count DESC;
```
